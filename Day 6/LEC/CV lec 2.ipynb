## Controlling vertical , horizontal line and point using trackbars

```python
import cv2
import numpy as np

def nothing(x):
    pass

img = np.zeros((300, 500), np.uint8)
cv2.namedWindow('controller')

cv2.createTrackbar('row', 'controller', 0, len(img)-1, nothing)
cv2.createTrackbar('col', 'controller', 0, len(img[0])-1, nothing)
cv2.createTrackbar('x', 'controller', 0, len(img)-1, nothing)
cv2.createTrackbar('y', 'controller', 0, len(img[0])-1, nothing)

while True:
    r = cv2.getTrackbarPos('row', 'controller')
    c = cv2.getTrackbarPos('col', 'controller')
    x = cv2.getTrackbarPos('x', 'controller')
    y = cv2.getTrackbarPos('y', 'controller')
    
    img[r,:]=255
    img[:,c]=255
    img[x,y]=255
    
    cv2.imshow('controller', img)

    img = np.zeros((300, 500), np.uint8)
    key = cv2.waitKey(10) 
    if key == ord('q'):
        break

        
cv2.destroyAllWindows()
```

## Drawing horizontal red color gradient line

```python
import cv2
import numpy as np

img = np.zeros((255, 255,3), np.uint8)
grad = (np.arange(0,len(img[0])))

for i in range(len(img[0])):
    img[50,i,:] = [0,0,grad[i]]
    
cv2.imshow('image', img)
key = cv2.waitKey(2000)
cv2.destroyAllWindows()

#img[50,:,0]= grad
#cv2.imshow('image', img)
#k = cv2.waitKey(200)
#cv2.destroyAllWindows()
```

## Drawing horizontal and vertical lines with random locations and random colors and then saving the result video

```python
import cv2
import numpy as np
import random

cv2.namedWindow('random')

frame_size = (255,255)

out = cv2.VideoWriter('output_vid.mp4',cv2.VideoWriter_fourcc(*'MP4V'),5.0,frame_size,True)

while True:
    img = np.zeros((255, 255,3), np.uint8)
    
    r = random.randint(0, len(img)-1)
    c = random.randint(0, len(img[0])-1)
    colorR = random.randint(0,255)
    colorC = random.randint(0,255)
        
    img[r,:,:]=np.random.randint(0, 255, size=3) 
    img[:,c,:]=np.random.randint(0, 255, size=3)
    
    cv2.imshow('random', img)
    out.write(img)
    
    key = cv2.waitKey(200) 
    if key == ord('q'):
        break

out.release()
cv2.destroyAllWindows()
```
