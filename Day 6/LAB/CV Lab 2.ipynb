## 2) run video stream using your laptop camera and save frames when i press key “s” close camera when i press “q” ,name of images should be “img.png +{counter}”


```python
import cv2
import numpy as np

cv2.namedWindow('video_frame_capture')

cap = cv2.VideoCapture(0)
cnt= 0

while(cap.isOpened()):
    state,frame = cap.read()
    if state == True:
        cv2.imshow('video_frame_capture',frame)
        key = cv2.waitKey(10)
        if key == ord('q'):
            break
        elif key == ord('s'):
            cnt=cnt+1
            cv2.imwrite(f'img{cnt}.png',frame)
    else:
        break

cap.release()
cv2.destroyAllWindows()
```

## 3) Make 100*100 rgb pixel image with black background and make diagonal of image with (x) color such that x is came from rgb trackbar

```python
import cv2
import numpy as np

def nothing(x):
    pass

cv2.namedWindow('controller')
cv2.createTrackbar('R', 'controller', 0, 255, nothing)
cv2.createTrackbar('G', 'controller', 0, 255, nothing)
cv2.createTrackbar('B', 'controller', 0, 255, nothing)

while True:
    img = np.zeros((100, 100,3), np.uint8)

    R = cv2.getTrackbarPos('R', 'controller')
    G = cv2.getTrackbarPos('G', 'controller')
    B = cv2.getTrackbarPos('B', 'controller')

    for i in range(min(len(img),len(img[0]))):
        img[i,i,:]=[B,G,R]

    cv2.imshow('controller', img)

    key = cv2.waitKey(10)
    if key == ord('q'):
        break
    elif key == ord('s'):
        cv2.imwrite('img.png',img)


cv2.destroyAllWindows()
```

## 4) create 100*100 rgb image with black background and make 1 vertical line and 1 horizontal line the color of both should be controlled spereat from 2 trackbar
## First one “horz” and second one is “vert”

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def nothing(x):
    pass

cv2.namedWindow('controller')
cv2.createTrackbar('R_vert', 'controller', 0, 255, nothing)
cv2.createTrackbar('G_vert', 'controller', 0, 255, nothing)
cv2.createTrackbar('B_vert', 'controller', 0, 255, nothing)
cv2.createTrackbar('R_horz', 'controller', 0, 255, nothing)
cv2.createTrackbar('G_horz', 'controller', 0, 255, nothing)
cv2.createTrackbar('B_horz', 'controller', 0, 255, nothing)

while True:
    img = np.zeros((100, 100 ,3), np.uint8)

    R_vert = cv2.getTrackbarPos('R_vert', 'controller')
    G_vert = cv2.getTrackbarPos('G_vert', 'controller')
    B_vert = cv2.getTrackbarPos('B_vert', 'controller')
    R_horz = cv2.getTrackbarPos('R_horz', 'controller')
    G_horz = cv2.getTrackbarPos('G_horz', 'controller')
    B_horz = cv2.getTrackbarPos('B_horz', 'controller')

    img[50,:,:]=[B_horz,G_horz,R_horz]
    img[:,50,:]=[B_vert,G_vert,R_vert]

    cv2.imshow('controller', img)

    key = cv2.waitKey(10)
    if key == ord('q'):
        break
    elif key == ord('s'):
        cv2.imwrite('img.png',img)



cv2.destroyAllWindows()
```

## 5) create 250*250 rgb image with black background and make 1 vertical line and 1 horizontal line with red color and both should be controlled spereat from 2 trackbar to select the horizontal line in which row and vertical line in which column.
## First one “horz” and second one is “vert” (Trackbar)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def nothing(x):
    pass

img = np.zeros((250, 250 ,3), np.uint8)

cv2.namedWindow('controller')
cv2.createTrackbar('horz', 'controller', 0, len(img)-1, nothing)
cv2.createTrackbar('vert', 'controller', 0, len(img[0])-1, nothing)

while True:
    img = np.zeros((250, 250 ,3), np.uint8)

    r = cv2.getTrackbarPos('horz', 'controller')
    c = cv2.getTrackbarPos('vert', 'controller')
    img[r,:]=[0,0,255]
    img[:,c]=[0,0,255]

    cv2.imshow('controller', img)

    key = cv2.waitKey(1)
    if key == ord('q'):
        break
    elif key == ord('s'):
        cv2.imwrite('img.png',img)


cv2.destroyAllWindows()
```

## 6) Read from camera and make video with 30 F/s

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

cv2.namedWindow('video_saver')

cap = cv2.VideoCapture(0)

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
frame_size = (width,height)

out = cv2.VideoWriter('output_vid.mp4',cv2.VideoWriter_fourcc(*'mp4v'),30.0,frame_size,True)
time = 0
while(cap.isOpened()):
    state,frame = cap.read()
    if state == True:
        cv2.imshow('video_saver',frame)
        key = cv2.waitKey(1)
        if key == ord('q'):
            break
        out.write(frame)
    else:
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```
