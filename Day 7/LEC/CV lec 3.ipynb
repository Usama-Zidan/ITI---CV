## Resizing image

```python
import cv2
import numpy as np 
import matplotlib.pyplot as plt 

img = cv2.imread('pic.jpg')
print(img.shape)

plt.imshow(img[:,:,::-1])

w = 300
h = 500

img2 = cv2.resize(img,(w,h),interpolation=cv2.INTER_CUBIC)
print(img2.shape)
plt.imshow(img2[:,:,::-1])
```

## Controlling line coordinates with Trackbars

```python
import cv2
import numpy as np 
import matplotlib.pyplot as plt

def nothing(x):
    pass

cv2.namedWindow('controller')
cv2.createTrackbar('x1', 'controller', 0, 499, nothing)
cv2.createTrackbar('y1', 'controller', 0, 499, nothing)
cv2.createTrackbar('x2', 'controller', 0, 499, nothing)
cv2.createTrackbar('y2', 'controller', 0, 499, nothing)

while True:
    img = np.zeros((500,500,3),np.uint8)

    x1 = cv2.getTrackbarPos('x1', 'controller')
    y1 = cv2.getTrackbarPos('y1', 'controller')
    x2 = cv2.getTrackbarPos('x2', 'controller')
    y2 = cv2.getTrackbarPos('y2', 'controller')
    
    cv2.line(img,(x1,y1),(x2,y2),(0,0,255),12)
    
    cv2.imshow('controller', img)
    
    key = cv2.waitKey(10) 
    if key == ord('q'):
        break
    elif key == ord('s'):
        cv2.imwrite('img.png',img) 
        
    
cv2.destroyAllWindows()
```

## Drawing radius (line from any point within circle circumference to center of the circle)

```python
import cv2
import numpy as np 
import matplotlib.pyplot as plt

img = np.zeros((500,500,3),np.uint8)
cv2.circle(img,(250,250),100,(0,0,255),5)

x2 = int(250 + 100 * np.cos(np.pi / 4))
y2 = int(250 - 100 * np.sin(np.pi / 4))

cv2.line(img, (250, 250), (x2, y2), (0, 0, 255), 5)

x2 = int(250 + 100 * np.cos(0))
y2 = int(250 - 100 * np.sin(0))

cv2.line(img, (250, 250), (x2, y2), (0, 0, 255), 5)

plt.imshow(img)
print()
```

## Drawing ploygon with connected ends

```python
import cv2
import numpy as np 
import matplotlib.pyplot as plt

img = np.zeros((500,500,3),np.uint8)
arr = np.array([[100,150],[250,250],[50,100],[350,150]],np.int32)

cv2.polylines(img,[arr],True,(0,0,255),2) 
plt.imshow(img)
```

## Drawing circle with a center point equals to point where you double click on window (Events)

```python
import cv2
import numpy as np 
import matplotlib.pyplot as plt 

img = np.zeros((500,500,3),np.uint8)

def callback(events,x,y,flags,param):
    if events == cv2.EVENT_LBUTTONDBLCLK:
        img = np.zeros((500,500,3),np.uint8)
        cv2.circle(img,(x,y),50,(0,0,255),5)
        cv2.imshow('image',img)
        
cv2.namedWindow('image')

cv2.setMouseCallback('image',callback)

cv2.imshow('image',img)
while True:
    key = cv2.waitKey(10)
    if key == ord('q'):
        break

cv2.destroyAllWindows()
```
