## 1) Create image 256 * 256 Draw red square 50*50 in the center of blue circle with radius 80 the center of both square and circle is the same center of image. 


```python
import cv2
import numpy as np

img = np.zeros((256, 256, 3), np.uint8)

cv2.circle(img, (128,128), 80, (255, 0, 0), -1)

cv2.rectangle(img, (128 - 25, 128 - 25), (128 + 25, 128 + 25), (0, 0, 255), -1)

cv2.imshow('image', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 2) make trackbar to control line position ,color and thickness in image 500*500 rgb.

```python
import cv2
import numpy as np

def nothing(x):
    pass

cv2.namedWindow('controller')

cv2.createTrackbar('x1', 'controller', 0, 499, nothing)
cv2.createTrackbar('y1', 'controller', 0, 499, nothing)
cv2.createTrackbar('x2', 'controller', 0, 499, nothing)
cv2.createTrackbar('y2', 'controller', 0, 499, nothing)
cv2.createTrackbar('R', 'controller', 0, 255, nothing)
cv2.createTrackbar('G', 'controller', 0, 255, nothing)
cv2.createTrackbar('B', 'controller', 0, 255, nothing)
cv2.createTrackbar('thickness', 'controller', 1, 15, nothing)

while True:
    img = np.zeros((500, 500, 3), np.uint8)
    
    x1 = cv2.getTrackbarPos('x1', 'controller')
    y1 = cv2.getTrackbarPos('y1', 'controller')
    x2 = cv2.getTrackbarPos('x2', 'controller')
    y2 = cv2.getTrackbarPos('y2', 'controller')
    R = cv2.getTrackbarPos('R', 'controller')
    G = cv2.getTrackbarPos('G', 'controller')
    B = cv2.getTrackbarPos('B', 'controller')
    thickness = cv2.getTrackbarPos('thickness', 'controller')

    cv2.line(img, (x1, y1), (x2, y2), (B, G, R), max(thickness,1))
    
    cv2.imshow('controller', img)
    key = cv2.waitKey(10)
    if key == ord('q'):
        break

        
cv2.destroyAllWindows()
```

## 3) make trackbar to control position,radius and color of circle in image 500*500 rgb.

```python
import cv2
import numpy as np

def nothing(x):
    pass

cv2.namedWindow('controller')

cv2.createTrackbar('x', 'controller', 10, 499, nothing)
cv2.createTrackbar('y', 'controller', 10, 499, nothing)
cv2.createTrackbar('radius', 'controller', 10, 250, nothing)
cv2.createTrackbar('R', 'controller', 0, 255, nothing)
cv2.createTrackbar('G', 'controller', 0, 255, nothing)
cv2.createTrackbar('B', 'controller', 0, 255, nothing)


while True:
    img = np.zeros((500, 500, 3), np.uint8)
    x_cor_center = cv2.getTrackbarPos('x', 'controller')
    y_cor_center = cv2.getTrackbarPos('y', 'controller')
    radius = cv2.getTrackbarPos('radius', 'controller')
    R = cv2.getTrackbarPos('R', 'controller')
    G = cv2.getTrackbarPos('G', 'controller')
    B = cv2.getTrackbarPos('B', 'controller')

    cv2.circle(img, (x_cor_center , y_cor_center), radius, (B,G,R), -1)
    cv2.imshow('controller', img)
    
    key = cv2.waitKey(10)
    if key == ord('q'):
        break
        

cv2.destroyAllWindows()
```

## 4) create image 800*800 rgb write a word “Hello”with red color and draw yellow rectangle on border of word

```python
import cv2
import numpy as np

img = np.zeros((800, 800, 3), np.uint8)

cv2.putText(img, 'Hello', (200, 200), cv2.FONT_ITALIC, 2, (0, 0, 255), thickness=4)

text_size, _ = cv2.getTextSize('Hello', cv2.FONT_ITALIC, 2, thickness=4)

cv2.rectangle(img, (200+4, 200 - text_size[1]), (200-7 + text_size[0], 200 + 3), (0, 255, 255), thickness=1)

cv2.imshow('image', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 5) create image 800*800 rgb with red circle in center of image with radius 100 and make small blue circle with radius of 6 that move within circumference of the big circle every 1 second 10 degree.

```python
import cv2
import numpy as np
import math


angle = 0

while True:
    img = np.zeros((800, 800, 3), np.uint8)
    cv2.circle(img, (400,400), 100, (0, 0, 255), -1)
    
    x = int(400 + 100 * np.cos(angle * np.pi / 180))
    y = int(400 - 100 * np.sin(angle * np.pi / 180))

    cv2.circle(img, (x, y), 6, (255, 0, 0), -1)
    angle += 10
    cv2.imshow('image', img)

    key = cv2.waitKey(1000)
    if key == ord('q'):
        break


cv2.destroyAllWindows()
```
