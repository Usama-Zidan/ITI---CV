```python
import numpy as np
import cv2
import random
    
img = np.zeros((500,500,3),np.uint8)

snake_pos = [[250,250]]
apple_pos = [random.randint(1,49)*10,random.randint(1,49)*10]
snake_head = [250,250]
score = 0

while True:

    cv2.imshow('Snake',img)

    if cv2.waitKey(100) == ord('q'):
        break
    
    img = np.zeros((500,500,3),np.uint8)

    cv2.putText(img, f"score : {score}", (20,50), cv2.FONT_ITALIC, .5, (0, 0, 255), 2)

    cv2.circle(img,(apple_pos[0],apple_pos[1]),2,(0,0,255),2)

    for position in snake_pos:
        cv2.circle(img,(position[0],position[1]),2,(0,255,0),2)
    
    if snake_head in snake_pos[1:]:
        img = np.zeros((500,500,3),np.uint8)
        cv2.putText(img, "Failed, Press any key to retry or q to exit", (40,250), cv2.FONT_ITALIC, .5, (0, 0, 255), 2)
        cv2.putText(img, f"score is : {score}", (150,300), cv2.FONT_ITALIC, .5, (0, 0, 255), 2)
        cv2.imshow('Snake',img)
        snake_pos = [[250,250]]
        score = 0
        key = cv2.waitKey(0)
        if key == ord('q'):
            break

    if(snake_head[0]>apple_pos[0]):
        snake_head[0] =  snake_head[0]-10
    elif(snake_head[0]<apple_pos[0]):
        snake_head[0] =  snake_head[0]+10
    elif(snake_head[1]<apple_pos[1]):
        snake_head[1] =  snake_head[1]+10
    elif(snake_head[1]>apple_pos[1]):
        snake_head[1] =  snake_head[1]-10    

    if snake_head == apple_pos:
        #increasing size
        apple_pos = [random.randint(1,49)*10,random.randint(1,49)*10]
        snake_pos.insert(0,list(snake_head))
        score = score +1

    else:
        #moving
        snake_pos.insert(0,list(snake_head))
        snake_pos.pop()
            
        
cv2.destroyAllWindows()
```
