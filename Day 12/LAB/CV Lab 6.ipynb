## 1) Build sequential model for equation:
## Z = 5 * X – 3 * Y + 7
## Task:
## Given the previous equation:
## - Generate 10 random values for `x` and `y` as 
## input with max. value = 100.
## - Generate z for the above values.
## - Create a model with 2 inputs to accept x and y 
## and predicts z.
## - You must show the results:
## - Model summary.
## - Loss VS epochs.
## - Weights.
## - Predict output for 3 different inputs.
## Criteria:
## - Model must be 1 neuron only.
## - Acc. must be 90% at least

```python
import numpy as np
import tensorflow as tf
import time
```

```python
input_data = np.random.randint(-100 , 100 , (10,2))
x = input_data[:,0]
y = input_data[:,1]
z = x * 5 - 3 * y + 7
```

```python
model = tf.keras.Sequential(
    tf.keras.layers.Dense(units=1, input_shape=[2])
)
```

```python
model.compile(
        loss='mean_squared_error',
        optimizer=tf.keras.optimizers.Adam(0.1),
    )
```

```python
model.summary()
```

```python
t = time.time()
hist = model.fit(input_data, z, epochs=1000)
print (str.format("Training took: {:0.2f} Sec", time.time() - t))
```

```python
import matplotlib.pyplot as plt
plt.xlabel('Number of Epoch')
plt.ylabel("Loss Value")
plt.plot(hist.history['loss']);
```

```python
print('Weights' , model.get_weights())
```

```python
input_test = np.random.randint(-100 , 100 , (3,2))
print('test input : \n' ,input_test)
print('test output : \n' ,model.predict(input_test))
```
