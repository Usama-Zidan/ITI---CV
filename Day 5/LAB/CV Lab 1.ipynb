## 1) read “Task_in” image and then print image shape ,plot the image using matplotlib and opencv then edit the right corner pixel to be red color then save the image as (.jpg,.npg) to your current folder name it “my_image.png” then read both image again. “Make your observation”.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = cv2.imread('Task_in.jpg')#BGR
print("Image shape:", img.shape)

plt.imshow(img)
plt.show()
cv2.imshow('image_raw',img)
cv2.waitKey(3000)
cv2.destroyAllWindows()


img[-1,-1] = [0, 0, 255] #BGR

cv2.imwrite('my_image.jpg', img)
cv2.imwrite('my_image.png', img)

img_jpg = cv2.imread('my_image.jpg')
img_png = cv2.imread('my_image.png')

print(img_jpg)
plt.imshow(img_jpg)
plt.show()
cv2.imshow('image jpg',img_jpg)
cv2.waitKey(3000)
cv2.destroyAllWindows()

print(img_png)
plt.imshow(img_png)
plt.show()
cv2.imshow('image png',img_png)
cv2.waitKey(3000)
cv2.destroyAllWindows()
```

## 2) make 51 * 51 * 3 RGB image with Blue color then save it as”Blue.jpg”

```python
img = np.zeros((51, 51, 3), np.uint8)

img[:, :, 0] = 255

cv2.imwrite("Blue.jpg", img)
```

## 3) Make trackbar for rgb colors with window “colors” 

```python
def nothing(x):
    print(x)
    
img = np.zeros((300, 500, 3), np.uint8)
cv2.namedWindow('colors')

cv2.createTrackbar('R', 'colors', 0, 255, nothing)
cv2.createTrackbar('G', 'colors', 0, 255, nothing)
cv2.createTrackbar('B', 'colors', 0, 255, nothing)

while True:
    cv2.imshow('colors', img)
    k = cv2.waitKey(1)
    if k == ord('q'):
        break

    r = cv2.getTrackbarPos('R', 'colors')
    g = cv2.getTrackbarPos('G', 'colors')
    b = cv2.getTrackbarPos('B', 'colors')
    img[:] = [b, g, r]

cv2.destroyAllWindows()
```
