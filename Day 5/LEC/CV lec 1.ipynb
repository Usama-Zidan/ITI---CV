```python
cap = cv2.VideoCapture('11_output.avi')

while(cap.isOpened()):
    state,frame = cap.read()
    if state == True:
        cv2.imshow('frame',frame)
        key = cv2.waitKey(1000)
        if key == ord('q'):
            break
        elif key == ord('s'):
            cv2.imwrite('out.jpg',frame)
    else:
        break

cap.release()
img = cv2.imread('out.jpg') #BGR
img = img[:,:,::-1] #RGB
plt.imshow(img)
cv2.destroyAllWindows()
```
