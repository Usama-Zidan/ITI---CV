## 2) Write a python code to find Sum and Average of N Natural Numbers

```python
num = int(input("Please Enter any number : "))
sum = 0
avg = 0
for i in range(num):
    sum = sum + (i+1)
avg = sum / num
print(f"The sum of natural numbers from 1 to {num} = {sum}")
print(f"The average of natural numbers from 1 to {num} = {avg}")
```

## 3) Python Program to Check Even or Odd

```python
exit_flag = False
while not exit_flag :
    exit = input('Enter \'q\' for exit.')
    if(exit.lower() == 'q'):
        break
        
    num = int(input('Enter any number : '))
    if(num%2):
        print(f"{num} is odd number")
    else:
        print(f"{num} is even number")
```

## Input Format

```python
string = input("enter a string : ")
print(string.swapcase()) 

#### another solution 

string2 = input("enter a string : ")
swap = ''
for char in string2:
    if char.upper() != char:
        swap += char.upper()
    elif char.lower() != char:
        swap += char.lower()
    else:
        swap += char
print(swap)
```

## 7) Implement a code that takes as an input ASCII code for 
## some characters and combine them to form a string.


```python
ascii_codes = input("Enter ascii codes separated by spaces: ")
ascii_codes = [int(code) for code in ascii_codes.split()]
string = ''.join([chr(char) for char in ascii_codes])
print(string)
```

## 9) Write a Python program which accepts the user's first and 
## last name and print them in reverse order with a space between them

```python
first_name = input("Enter your first name: ")
last_name = input("Enter your last name: ")
full_name = first_name + " " + last_name

reverse = ""
for i in full_name:
    reverse = i + reverse
print(reverse)
```
