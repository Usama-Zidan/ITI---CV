## 1) load the following image and make red rectangle around face and write below rectangle “Cat” with yellow color.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 

img = cv2.imread('original.png')

cv2.rectangle(img,(780,120),(1480,820),(0,0,255),12)

cv2.putText(img,'Cat' , (1100,880),cv2.FONT_ITALIC,2,(0,255,255),4,cv2.LINE_AA)

plt.imshow(img[:,:,::-1])
```

## 2)  load the following image and apply sharpen filter.“write your observation"

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 

src = cv2.imread('blurred.png')

ker = np.array(
    [
        [-1,-1,-1],
        [-1,9,-1],
        [-1,-1,-1]
    ]
)

out = cv2.filter2D(src , -1,ker)

f, (ax0,ax1) = plt.subplots(1,2,figsize=(20,20))

ax0.set_title("Blurred")
ax0.imshow(src[:,:,::-1])

ax1.set_title("Sharpened")
ax1.imshow(out[:,:,::-1])
```

## Observation on Q2 : 
## - the image became a little more definite but it's still blurry and unclear
## - darker areas became more obivous but it didn't make the photo much better

## 3)  load the following image and apply blur filter “write your observation”

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 

src = cv2.imread('sharpen.png')

ker_avg = np.array([
        [1,1,1],
        [1,1,1],
        [1,1,1]
    ])/9

avg = cv2.filter2D(src , -1,ker_avg)
avg2 = cv2.blur(src,(11,11))

f, (ax0,ax1,ax2) = plt.subplots(1,3,figsize=(20,20))

ax0.set_title("Sharpenend")
ax0.imshow(src[:,:,::-1])

ax1.set_title("Average filter 3*3")
ax1.imshow(avg[:,:,::-1])

ax2.set_title("Average filter 5*5")
ax2.imshow(avg2[:,:,::-1])
```

## Observation on Q3 : 
## - the image is still clear on 3 * 3 average filter kernel 
## - the whites became less , it can be seen in the hair or fur
## - using a bigger kernel made it lose more details as seen in 11 * 11 kernel and increasing it more will make it unrecognisable
