```python
import cv2
import numpy as np
import datetime
import math

while True:
    img = np.ones((1000, 1000, 3), np.uint8)*255

    cv2.circle(img, (500, 500), 310, (0,0,0), 8)

    for i in range(0, 60):
        x = int(500 + 305 * np.cos(np.pi * (i) / 30))
        y = int(500 - 305 * np.sin(np.pi *(i) / 30))
        cv2.circle(img, (x,y),4,(0,0,255),-1)

    for i in range(0, 12):
        x = int(500 + 305 * np.cos(np.pi * (i) / 6))
        y = int(500 + 305 * np.sin(np.pi *(i) / 6))
        cv2.circle(img, (x,y),9,(0,0,255),-1)

    loc = [(660, 220), (780, 335),(830, 510),(785, 675), (660, 805), (490, 850), (320, 805), (190, 680), (150, 510), (180, 330), (310, 210), (480, 160)]
    for i in range(0, 12):
        cv2.putText(img, f"{i+1}", (loc[i][0], loc[i][1]), cv2.FONT_ITALIC, 1, (0, 0, 0), 2)

    time_now = datetime.datetime.now().time()
    date_now = datetime.datetime.now().date()
    hour = time_now.hour % 12
    minute = time_now.minute
    sec = time_now.second
    micro_sec = time_now.microsecond

    current_time = "Time : " + time_now.strftime("%H:%M:%S")
    current_date = "Date : " + date_now.strftime("%d-%m-%Y")
    cv2.putText(img, current_time, (50, 100), cv2.FONT_ITALIC, 1, (0, 0, 0), 2)
    cv2.putText(img, current_date, (50, 50), cv2.FONT_ITALIC, 1, (0, 0, 0), 2)

    sec_angle = sec * 6 + micro_sec / 1000000 * 6
    min_angle = minute * 6 + sec / 10
    hr_angle = (hour % 12 )* 30 + (minute / 2) + (sec / 120)

    sec_x = int(500 + 310 * math.cos(math.radians(sec_angle)-22/14))
    sec_y = int(500 + 310 * math.sin(math.radians(sec_angle)-22/14))
    cv2.line(img, (500, 500), (sec_x, sec_y), (0,0,0), 2)

    min_x = int(500 + 260 * math.cos(math.radians(min_angle)-22/14))
    min_y = int(500 + 260 * math.sin(math.radians(min_angle)-22/14))
    cv2.line(img, (500, 500), (min_x, min_y), (0,0,0), 8)

    hr_x = int(500 + 180 * math.cos(math.radians(hr_angle)-22/14))
    hr_y = int(500 + 180 * math.sin(math.radians(hr_angle)-22/14))
    cv2.line(img, (500, 500), (hr_x, hr_y), (0,0,0), 10)

    cv2.circle(img, (500, 500), 10, (0,0,0), -1)

    cv2.putText(img, "Press q to exit", (400, 950), cv2.FONT_ITALIC, 1, (0, 0, 0), 2)

    cv2.imshow("clock", img)

    key = cv2.waitKey(1)
    if key == ord('q'):
        break

cv2.destroyAllWindows()
```
